```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5]
y1 = [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]
y2 = [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]
y3 = [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73]
x4 = [8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8]
y4 = [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]
```

```python
m, b = np.polyfit(x, y1, 1)
regression_line = m * np.array(x) + b
```

```python
plt.scatter(x, y1)
plt.plot(x, regression_line, color='red')
plt.xlabel('x')
plt.ylabel('y1')
plt.title('x vs y1')
plt.show()
```

```python
m2, b2 = np.polyfit(x, y2, 1)
regression_line2 = m2 * np.array(x) + b2

plt.scatter(x, y2, color='blue')
plt.plot(x, regression_line2, color='red')
plt.xlabel('x')
plt.ylabel('y2')
plt.title('x vs y2')
plt.show()
```

```python
m3, b3 = np.polyfit(x, y3, 1)
regression_line3 = m3 * np.array(x) + b3

plt.scatter(x, y3, color='green')
plt.plot(x, regression_line3, color='red')
plt.xlabel('x')
plt.ylabel('y3')
plt.title('x vs y3')
plt.show()
```

```python
m4, b4 = np.polyfit(x4, y4, 1)
regression_line4 = m4 * np.array(x4) + b4

plt.scatter(x4, y4, color='orange')
plt.plot(x4, regression_line4, color='red')
plt.xlabel('x4')
plt.ylabel('y4')
plt.title('x4 vs y4')
plt.show()
```

#Reflection Questions

### How do the datasets visually differ despite having similar summary statistics?


The four data sets presented have very similar mean, variance, and correlations, but when visualized on a graph they are presented quite differently. Y1 is presented in the typical straight line relationship. Y2 is curved and not straight. Y3 is straight, just like Y1, but one point is farther away from the rest of the points. X4 has the most different presentation as all points are at the same number, however one point is far away from the rest. These all present different variations on how to interpret the data presented as none of them are the same as the graph shows the relationship differently.


### How did the outlier in the x4, y4 dataset affect the regression line compared to the other datasets?


In X4, the value for most point is 8 except for the one point that sits a 19. This point is the reason the slope is made. Without this number all points would be sitting on top of each other with no real trend being showcased, thats why the 19 or outlier is there to create a relationship.


### Why is it important to visualize data in addition to calculating summary statistics?


It is important to visualize data because while data can look similar, often times when presented on a graph the relationships and way the data has a relationship is different in each graph. Graphing allows for data to be properly understood, not just by looking at the numbers and trying to figure out what the data means.
