```python
import numpy as np
import matplotlib.pyplot as plt
```

#PART 1 ***Scatter Plot***

```python
study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]

plt.figure(figsize=(6,4))
plt.scatter (study_hours, exam_scores, marker='o', color='Purple')
plt.title("Time Spent Studying vs Final Score")
plt.xlabel("Time Spent Studying")
plt.ylabel("Final Score")
plt.grid(alpha=0.3)
plt.show()
```

There is a positive relationship between Time Spent Studying and Final score because as Time Spent Studying increases so does Final Score

# Part 2 ***Line Plot***

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
city_a = [68, 70, 72, 71, 74, 76, 75]
city_b = [62, 64, 67, 66, 69, 71, 70]

plt.figure(figsize=(7,4))
plt.plot(days, city_a, marker='o', linestyle='-', label='City A')
plt.plot(days, city_b, marker='s', linestyle='--', label='City B')
plt.title("Temperture During One Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

City A is the Warmer city compared to City B as City A's Line sits about City B's Line at all days of the week

#Part 3A: ***Basic Bar Chart***

```python
offenses = ["Aggravated\nAssault", "Burglary", "Larceny", "Motor Vehicle\nTheft",
            "Murder", "Rape", "Robbery"]
rates_2023 = [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]

plt.figure(figsize=(9,5))
plt.bar(offenses, rates_2023, color='Green')
plt.title("Crime Rates in 2023 per 100,000 People")
plt.xlabel("Offense Type")
plt.ylabel("Rate per 100,000 People")
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

#Part 3B: ***Horizontal Bar Chart — % Change***

```python
offenses = ["Aggravated Assault", "Burglary", "Larceny", "Motor Vehicle Theft",
            "Murder", "Rape", "Robbery"]
rates_2023 = [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]
change = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]

rates_2022 = [r - c for r, c in zip(rates_2023, change)]
pct_change = [(c / r2022) * 100 for c, r2022 in zip(change, rates_2022)]

colors = ['Red' if x > 0 else 'Gold' for x in pct_change]

plt.figure(figsize=(8,5))
plt.barh(offenses, pct_change, color=colors)
plt.axvline(0, color='black', linewidth=0.8)
plt.title("% Change in Crime Rates (2022–2023)")
plt.xlabel("Percent Change")
plt.ylabel("Offense Type")
plt.tight_layout()
plt.show()
```

Part 3A showcases the size of each crime rate, so catergories like larceny are going to stand out more even if they didn't change that much in Part 3B. Part 3B is showcasing the amount of change based on % to see which crimes are going up and down, like Motor Vehicle Theft going up or murder going down

#Part 3C: ***Improved Version***

```python
sorted_data = sorted(zip(offenses, pct_change), key=lambda x: x[1])
sorted_offenses, sorted_pct = zip(*sorted_data)
colors_sorted = ['Red' if x > 0 else 'Gold' for x in sorted_pct]

plt.figure(figsize=(9,5.5))
bars = plt.barh(sorted_offenses, sorted_pct, color=colors_sorted)
plt.axvline(0, color='black', linewidth=0.8)
plt.title("% Change in Crime Rates (2022–2023)", fontsize=14)
plt.xlabel("Percent Change", fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.4)

# Add data labels
for bar, val in zip(bars, sorted_pct):
    plt.text(val + (0.5 if val > 0 else -0.5), bar.get_y() + bar.get_height()/2,
              f"{val:.1f}%", va='center', ha='left' if val > 0 else 'right', fontsize=10)

plt.tight_layout()
plt.show()
```

#Part 3D: ***Grouped Bar Chart***

```python
companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

x = np.arange(len(companies))
width = 0.25

plt.figure(figsize=(7,5))
plt.bar(x - width, product1, width, label='Product 1')
plt.bar(x, product2, width, label='Product 2')
plt.bar(x + width, product3, width, label='Product 3')

plt.xticks(x, companies)
plt.title("Sales by Company and Product")
plt.xlabel("Company")
plt.ylabel("Sales")
plt.legend()
plt.tight_layout()
plt.show()
```
